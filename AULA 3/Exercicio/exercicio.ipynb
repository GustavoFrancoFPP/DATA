```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

df = pd.read_csv('super store.csv')

display(df.head())
df.info()

plt.figure(figsize=(10,6))
sns.kdeplot(data=df, x='Sales', fill=True, color='tomato')
plt.title('Distribuição de vendas')
plt.xlabel('Vendas')
plt.ylabel('Densidade')
plt.show()

plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Region', y='Sales', palette='Set2')
plt.title('Preço da venda por região')
plt.xlabel('Região')
plt.ylabel('Vendas')
plt.show()

plt.figure(figsize=(10,6))
sns.kdeplot(data=df, x='Sales', hue='Ship Mode', fill=True, common_norm=False, palette='crest')
plt.title('Densidade de vendas por modo de envio')
plt.xlabel('Vendas')
plt.show()

plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Profit', hue='Segment', multiple='stack', bins=25, palette='pastel')
plt.title('Distribuição de lucro por segmento')
plt.xlabel('Lucro')
plt.show()

plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Region', y='Profit', hue='Segment', palette='Set3')
plt.title('Impacto do segmento no lucro por região')
plt.xlabel('Região')
plt.ylabel('Lucro')
plt.legend(title='Segmento')
plt.show()

plt.figure(figsize=(10,6))
sns.violinplot(data=df, x='Region', y='Sales', hue='Segment', split=True, palette='muted')
plt.title('Distribuição de vendas por região e segmento')
plt.xlabel('Região')
plt.ylabel('Vendas')
plt.show()
```
